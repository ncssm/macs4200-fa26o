# Activity 30: Diffie–Hellman Key Exchange

Diffie–Hellman is only as secure as the **discrete log problem**: Eve knows $p$, $g$ and Alice's public key $A = g^a \pmod{p}$, but finding Alice's private exponent $a$ from them is very hard. The most obvious attack is brute force: try $a = 1, 2, 3, \ldots$ until $g^a \pmod{p}$ equals $A$.

In this notebook you'll time that brute-force search for primes of different sizes, and record the times in the table on your Activity 30 handout. Then you'll use your times to estimate how long the search would take on a real Diffie–Hellman prime.

## The brute-force search

`crack_diffie_hellman(g, p, A)` tries every value of $a$ from $1$ to $p - 1$ and returns the first one where $g^a \equiv A \pmod{p}$. If no value works, it returns `None`. Run the cell below so the function is ready to use.

```python
def crack_diffie_hellman(g, p, A):
    for a in range(1, p):
        if pow(g, a, p) == A:
            return a
    return None   # no exponent works: g never reaches A
```

## A small example

In question 1 of the handout, Alice and Bob use $p = 29$ and $g = 2$, and Alice picks $a = 5$. Her public key is $2^5 \bmod 29 = 3$. Eve only sees $p = 29$, $g = 2$ and $A = 3$. The cell below plays Eve and finds $a$.

```python
crack_diffie_hellman(2, 29, 3)
```

## Timing a search

Python can time a line of code with `%timeit`. The two settings control how many times it runs:

- `-n` is how many times the search runs in one batch.
- `-r` is how many batches it runs.

`%timeit` reports the average time for **one** search, plus or minus how much the batches varied. Watch the units: ns (nanoseconds), µs (microseconds), ms (milliseconds) or s (seconds).

The cell below times the first row of the handout's table: $p = 29$, $g = 2$, $A = 5$.

```python
%timeit -n100 -r5 crack_diffie_hellman(2, 29, 5)
```
