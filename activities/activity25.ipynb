```python
def get_bit(value, bit_index):
    bitmask = 1 << bit_index
    result = value & bitmask
    return result >> bit_index

def set_bit(value, bit_index):
    bitmask = 1 << bit_index
    result = value | bitmask
    return result 

def clear_bit(value, bit_index):
    bitmask = 1 << bit_index
    result = value & ~bitmask
    return result
```

```python
def lfsr5(seed, num_bits):
    current_state = seed
    output = 0
    
    for i in range(num_bits):
        
        output_bit = get_bit(current_state, 0)
        shifted_output = output << 1
        
        if output_bit == 1:
            output = set_bit(shifted_output, 0)
        else:
            output = clear_bit(shifted_output, 0)
        
        new_bit = get_bit(current_state, 4) ^ get_bit(current_state, 1) ^ get_bit(current_state, 0)
        
        shifted_current_state = current_state >> 1
        
        if new_bit == 1:
            current_state = set_bit(shifted_current_state, 4)
        else:
            current_state = clear_bit(shifted_current_state, 4)
    
    return output
```

```python
def lfsr4(seed, num_bits):
    current_state = seed
    output = 0
    
    for i in range(num_bits):
        
        output_bit = get_bit(current_state, 0)
        shifted_output = output << 1
        
        if output_bit == 1:
            output = set_bit(shifted_output, 0)
        else:
            output = clear_bit(shifted_output, 0)
        
        new_bit = get_bit(current_state, 2) ^ get_bit(current_state, 1) ^ get_bit(current_state, 0)
        
        shifted_current_state = current_state >> 1
        
        if new_bit == 1:
            current_state = set_bit(shifted_current_state, 3)
        else:
            current_state = clear_bit(shifted_current_state, 3)
    
    return output
```

```python
def lfsr3(seed, num_bits):
    current_state = seed
    output = 0
    
    for i in range(num_bits):
        
        output_bit = get_bit(current_state, 0)
        shifted_output = output << 1
        
        if output_bit == 1:
            output = set_bit(shifted_output, 0)
        else:
            output = clear_bit(shifted_output, 0)
        
        new_bit = get_bit(current_state, 1) ^ get_bit(current_state, 0)
        
        shifted_current_state = current_state >> 1
        
        if new_bit == 1:
            current_state = set_bit(shifted_current_state, 2)
        else:
            current_state = clear_bit(shifted_current_state, 2)
    
    return output
```
