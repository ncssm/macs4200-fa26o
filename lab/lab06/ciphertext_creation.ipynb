```python
def get_bit(value, bit_index):
    bitmask = 1 << (bit_index - 1) # SOLUTION
    result = value & bitmask  # SOLUTION
    return result >> (bit_index - 1)  # SOLUTION
```

```python
def set_bit(value, bit_index):
    bitmask = 1 << (bit_index - 1) # SOLUTION
    result = value | bitmask  # SOLUTION
    return result  # SOLUTION 
```

```python
def clear_bit(value, bit_index):
    bitmask = 1 << (bit_index - 1) # SOLUTION
    result = value & ~bitmask  # SOLUTION
    return result  # SOLUTION 
```

```python
def lfsr8(seed, num_bits):
    """
            8 7 6 5 4 3 2 1 (bits == 8)
           ┌─┬─┬─┬─┬─┬─┬─┬─┐
        ┌─→│1│0│1│0│0│0│0│1├─→ output
        │  └─┴─┴─┴┬┴┬┴┬┴─┴┬┘
        └──────XOR┘ │ │   │
                └─XOR └─┐ │
                    └──XOR┘
                          (taps == 5, 4, 3, 1)
    """
    current_state = seed
    output_stream = 0
    
    for i in range(num_bits):

        print(format(current_state, '08b'))
        
        # gets the value of the bit at $b_1$ in the current state
        output_bit = get_bit(current_state, 1) # SOLUTION
        
        # left-shift the output stream by 1
        output_stream = output_stream << 1  # SOLUTION
        
        # If the output bit is 1, set the rightmost bit in the output stream
        if output_bit == 1:
            output_stream = set_bit(output_stream, 1)  # SOLUTION
        
        # Get the value of bits 5, 4, 3, and 1 and **XOR** those values to determine value of the new bit
        new_bit = get_bit(current_state, 5) ^ get_bit(current_state, 4) ^ get_bit(current_state, 3) ^ get_bit(current_state, 1)  # SOLUTION
        
        # Right-shift the current state by 1, creating room for a new bit at position 8
        current_state = current_state >> 1  # SOLUTION
        
        # if the new_bit is 1, then set b_8 in shifted_current_state to 1 and assign to current_state
        if new_bit == 1:
            current_state = set_bit(current_state, 8)  # SOLUTION
    
    return output_stream
```

```python
def lfsr6(seed, num_bits):
    current_state = seed
    output_stream = 0
    
    for i in range(num_bits):
        output_bit = current_state & 1
        output_stream = output_stream << 1
        if output_bit == 1:
            output_stream = output_stream | 1   
        new_bit = (current_state & 2**(2))>>2 ^ (current_state & 2**(1))>>1 ^ (current_state & 2**(1-1))
        current_state = current_state >> 1
        if new_bit == 1:
            current_state = current_state | 2**(6-1)
   
    return output_stream
```
